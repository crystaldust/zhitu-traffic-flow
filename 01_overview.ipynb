# 交通流量综合案例实践

在现实世界中，城市交通在不同时段有着不同的繁忙程度，例如早晚高峰期，交通流量突然增加，其他时间则流量会变得比较小。但是，不同城市，或者同一城市中的不同路口，其交通流量都是不同的，会受到各种细微因素的影响，显然，仅仅针对固定时段的交通流量进行管理，无法获得最优结果。在人工智能领域中，不同时段的交通流量数据可以按照时序数据(time series data)的形式来处理，通过对以往交通流量数据的学习，可以更加精确、灵活的对未来交通流量进行动态化的预测。

为了完成一整套交通流量检测的解决方案，仅有时序数据预测模型还不够，我们需要收集交通流量数据。假设我们现在是一个解决方案提供商，我们应该如何去收集交通流量数据呢？我们假设可以访问路口交通摄像头的数据，那么，我们仅需要对出现在路口的机动车进行统计，就可以大概的获得交通流量数据，用人工的方式显然是非常低效的，我们可以使用已经学习到的人工智能技术，用模型去检测画面中的机动车车牌号码（因为车牌号码是唯一的），统计一段时间内，镜头中出现的车辆数量，以获取交通流量数据。

在这个案例中，我们将综合使用目标检测、OCR以及时序数据预测等AI技术，构建一个简单的交通流量预测的解决方案，让大家能够对已有的AI技术进行综合应用，进行端到端的综合实践。整个项目的结构如下图所示：
